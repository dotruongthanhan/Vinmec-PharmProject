Nạp ca DEMO-02, chạy đường chạy luật, lắp `Synthesizer` và kiểm tra đầu ra `KetQuaLLM`.

```python
import sys
from pathlib import Path

root = Path.cwd().resolve()
while not (root / "pyproject.toml").is_file() and root != root.parent:
    root = root.parent
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
```

```python
from app.container import build_container
from app.llm.setup import build_synthesizer
from app.pipeline import run_case

container = build_container()
ca = container.case_repo.get("DEMO-02")
ket_qua = run_case(ca, container.renal_rules)
synthesizer = build_synthesizer(container)
```

```python
ket_qua_llm = synthesizer.run(ca, ket_qua, doan=[])
print(f"Trạng thái: {ket_qua_llm.trang_thai}")
print(f"Số lần gọi: {ket_qua_llm.so_lan_goi}")
print(f"Phiên bản prompt: {ket_qua_llm.phien_ban_prompt}")
print(f"Mô hình: {ket_qua_llm.mo_hinh}")
print(f"Cảnh báo: {ket_qua_llm.canh_bao}")
print(f"last_usage: {getattr(synthesizer.llm, 'last_usage', None)}")
```

```python
for idx, item in enumerate(ket_qua_llm.nhan_dinh, start=1):
    if item.phat_hien is not None:
        yeu_to = item.phat_hien.yeu_to
        ket_luan = item.phat_hien.ket_luan
        nhan_xet = item.phat_hien.mo_ta
    elif item.de_xuat is not None:
        yeu_to = item.de_xuat.yeu_to
        ket_luan = item.de_xuat.ket_luan
        nhan_xet = item.de_xuat.nhan_xet
    else:
        yeu_to = None
        ket_luan = None
        nhan_xet = ""
    print(
        f"[{idx}] nguồn={item.nguon_goc} | yếu tố={yeu_to} | "
        f"kết luận={ket_luan} | lỗi={item.loi_kiem_chung}"
    )
    print(f"    nhận xét: {nhan_xet}")
    if item.dien_giai is not None:
        print(f"    diễn giải: {item.dien_giai.nhan_xet}")
```
